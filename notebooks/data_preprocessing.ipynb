## Data Preprocessing for Machine Learning
### Training Data

```python
# imports
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
import joblib
```

```python
raw_data_train = pd.read_csv('../data/Train.csv')
raw_data_train.head()
```

```python
# Encoding categorical variables
copy_data=raw_data_train.__deepcopy__()
ge=LabelEncoder()
em=LabelEncoder()
gr=LabelEncoder()
pr=LabelEncoder()
ss=LabelEncoder()
se=LabelEncoder()
va=LabelEncoder()
# Filling NA values with appropriate values
copy_data['Ever_Married'].fillna('No', inplace=True)
copy_data['Graduated'].fillna('No', inplace=True)
copy_data['Profession'].fillna('Other', inplace=True)
copy_data['Work_Experience'].fillna(0, inplace=True)
copy_data['Family_Size'].fillna(0, inplace=True)

copy_data['Gender'] = ge.fit_transform(copy_data['Gender'])
copy_data['Ever_Married'] = em.fit_transform(copy_data['Ever_Married'])
copy_data['Graduated'] = gr.fit_transform(copy_data['Graduated'])
copy_data['Profession'] = pr.fit_transform(copy_data['Profession'])
copy_data['Spending_Score'] = ss.fit_transform(copy_data['Spending_Score'])
copy_data['Segmentation'] = se.fit_transform(copy_data['Segmentation'])
copy_data['Var_1'] = va.fit_transform(copy_data['Var_1'])


# Saving the encoders
joblib.dump(ge, '../data/preprocessed_data/gender_encoder.pkl')
joblib.dump(em, '../data/preprocessed_data/married_encoder.pkl')
joblib.dump(gr, '../data/preprocessed_data/graduated_encoder.pkl')
joblib.dump(pr, '../data/preprocessed_data/profession_encoder.pkl')
joblib.dump(ss, '../data/preprocessed_data/spending_score_encoder.pkl')
joblib.dump(se, '../data/preprocessed_data/segmentation_encoder.pkl')
joblib.dump(va, '../data/preprocessed_data/var1_encoder.pkl')
```

```python
copy_data
```

```python
copy_data.isna().sum()
```

```python
# Dropping unnecessary columns
# important columns are: features=['Ever_Married', 'Age', 'Graduated', 'Profession', 'Family_Size','Spending_Score']

X_train=copy_data.drop(['Segmentation','ID','Gender','Work_Experience','Var_1'], axis=1)
y=copy_data['Spending_Score']
```

```python
# Normalizing the data
scaler = StandardScaler()
X_train_norm=scaler.fit_transform(X_train)
X_train_norm = pd.DataFrame(X_train_norm, columns=X_train.columns)

# Saving the X_train data for later use
joblib.dump(X_train_norm,'../data/preprocessed_data/X_train_norm.pkl')
```

## Test Data

```python
raw_data_test= pd.read_csv('../data/Test.csv')
raw_data_test
```

```python
raw_data_test.isna().sum()
```

```python
copy_test_data=raw_data_test.__deepcopy__()
# Filling NA values with appropriate values
copy_test_data['Ever_Married'].fillna('No', inplace=True)
copy_test_data['Graduated'].fillna('No', inplace=True)
copy_test_data['Profession'].fillna('Other', inplace=True)
copy_test_data['Work_Experience'].fillna(0, inplace=True)
copy_test_data['Family_Size'].fillna(0, inplace=True)

# Encoding categorical variables
le = LabelEncoder()
copy_test_data['Gender'] = le.fit_transform(copy_test_data['Gender'].map({'Male': 1, 'Female' : 0}))
copy_test_data['Ever_Married'] = le.fit_transform(copy_test_data['Ever_Married'].map({'Yes': 1, 'No' : 0}))
copy_test_data['Graduated'] = le.fit_transform(copy_test_data['Graduated'].map({'Yes': 1, 'No' : 0}))
copy_test_data['Profession'] = le.fit_transform(copy_test_data['Profession'].map({'Healthcare': 0, 'Entertainment' : 1, 'Other' : 2, 'Academic' : 3, 'Artist' : 4, 'Sports' : 5}))
copy_test_data['Spending_Score'] = le.fit_transform(copy_test_data['Spending_Score'].map({'Low': 0, 'Medium' : 1, 'High' : 2}))
copy_test_data['Segmentation'] = le.fit_transform(copy_test_data['Segmentation'].map({'A': 0, 'B' : 1, 'C' : 2, 'D' : 3}))
copy_test_data['Var_1'] = le.fit_transform(copy_test_data['Var_1'].map({'Cat_1': 0, 'Cat_2' : 1, 'Cat_3' : 2, 'Cat_4' : 3, 'Cat_5' : 4, 'Cat_6' : 5, 'Cat_7' : 6, 'Cat_8' : 7, 'Cat_9' : 8}))
```

```python
copy_test_data.isnull().sum()
```

```python
# Dropping unnecessary columns
X_test=copy_test_data.drop(['Segmentation','ID','Gender','Work_Experience','Var_1'], axis=1)
y=copy_test_data['Spending_Score']

# Normalizing the data
scaler = StandardScaler()
X_test_norm=scaler.fit_transform(X_test)
X_test_norm = pd.DataFrame(X_test_norm, columns=X_test.columns)

joblib.dump(scaler,'../data/preprocessed_data/scaler.pkl')
# Saving the scaler for later use

# Saving the X_test data for later use
joblib.dump(X_test_norm,'../data/preprocessed_data/X_test_norm.pkl')
```
